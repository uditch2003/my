```python
# high disc low prof
df['discount_group'] = pd.cut(
    df['discount'],
    bins=[-0.01, 0.10, 0.20, 0.30, 0.50, 1],
    labels=[
        '0-10%',
        '10-20%',
        '20-30%',
        '30-50%',
        '50%+'
    ]
)
discount_analysis = df.groupby(
    'discount_group',
    observed=True
).agg(
    sales=('sales', 'sum'),
    profit=('profit', 'sum'),
    avg_profit=('profit', 'mean'),
    orders=('order_id', 'nunique')
)

discount_analysis
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicates:")
print(df.duplicated().sum())

# 2. KPI
print("\n========== KPIs ==========")

print("Total Sales:", df['sales'].sum())
print("Total Profit:", df['profit'].sum())
print("Total Quantity:", df['quantity'].sum())
print("Total Orders:", df['order_id'].nunique())
print("Total Customers:", df['customer_id'].nunique())
print("Total Products:", df['product_id'].nunique())

profit_margin = (
    df['profit'].sum() /
    df['sales'].sum()
) * 100

print("Profit Margin:", round(profit_margin, 2), "%")


# 3. CATEGORY
print("\n========== CATEGORY ==========")

print(
    df.groupby('category')
      .agg(
          Sales=('sales', 'sum'),
          Profit=('profit', 'sum'),
          Quantity=('quantity', 'sum'),
          Orders=('order_id', 'nunique')
      )
      .sort_values('Sales', ascending=False)
)


# 4. REGION
print("\n========== REGION ==========")

print(
    df.groupby('region')
      .agg(
          Sales=('sales', 'sum'),
          Profit=('profit', 'sum'),
          Customers=('customer_id', 'nunique'),
          Orders=('order_id', 'nunique')
      )
      .sort_values('Sales', ascending=False)
)


# 5. TOP PRODUCTS
print("\n========== TOP PRODUCTS ==========")

print(
    df.groupby('product_name')['sales']
      .sum()
      .nlargest(10)
)


# 6. WORST PRODUCTS BY PROFIT
print("\n========== LOWEST PROFIT PRODUCTS ==========")

print(
    df.groupby('product_name')['profit']
      .sum()
      .nsmallest(10)
)


# 7. SEGMENT
print("\n========== CUSTOMER SEGMENT ==========")

print(
    df.groupby('segment')
      .agg(
          Sales=('sales', 'sum'),
          Profit=('profit', 'sum'),
          Customers=('customer_id', 'nunique')
      )
      .sort_values('Sales', ascending=False)
)


# 8. SHIPPING
print("\n========== SHIPPING ==========")

print(
    df.groupby('ship_mode')
      .agg(
          Orders=('order_id', 'nunique'),
          Avg_Shipping_Days=('shipping_days', 'mean')
      )
)


# 9. YEARLY SALES
print("\n========== YEARLY SALES ==========")

print(
    df.groupby('order_year')['sales']
      .sum()
)


# 10. YEARLY PROFIT
print("\n========== YEARLY PROFIT ==========")

print(
    df.groupby('order_year')['profit']
      .sum()
)
```
